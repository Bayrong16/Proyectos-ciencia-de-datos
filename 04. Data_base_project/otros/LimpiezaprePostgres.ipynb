Limpieza de datos realizada a algunos archivos .csv, antes de importarlos a PosgresSQL

#Limpieza tariffs

```python
import pandas as pd

# 1. Cargar el archivo CSV
# Reemplaza 'tariffs.csv' por la ruta de tu archivo original
df = pd.read_csv('tariffs.csv')

```

```python
df.head()
```

```python
# 2. Rellenar todos los valores vacíos (NaN) con 0
df = df.fillna(0)


# 4. Guardar el archivo corregido listo para importar
#df.to_csv('tariffs_limpio.csv', index=False)

#print("¡Archivo procesado con éxito! Se guardó como 'tariffs_limpio.csv'")
```

```python
df.head()
```

```python
df.info()
```

```python
df['cdi'] = df['cdi'].astype(int)
display(df.head())
df.info()
```

```python

# 4. Guardar el archivo corregido
df.to_csv('tariffs_limpio.csv', index=False)
```

#Limpieza services

```python
# 1. Cargar el archivo services.csv original
df_services = pd.read_csv('services.csv')
```

```python
df_services.head()
```

```python
# 2. Aplicar la regla de negocio: si voltage_level es 2 o 3, fijamos cdl en 0
# (para que coincida con lo que guardaste en tariffs)
df_services.loc[df_services['voltage_level'].isin([2, 3]), 'cdi'] = 0
```

```python
df_services.head()
```

```python
df_services.info()
```

```python
# 5. Guardar el nuevo archivo limpio
df_services.to_csv('services_limpio.csv', index=False)

print("¡Archivo services_limpio.csv creado con éxito!")
```
